# Palmer Penguins

## Analyze dataset

```python
# imports
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# load dataset
df = sns.load_dataset("penguins")

# observe dataset
print(f"Dataset shape: {df.shape}")
df.info()
df.describe()
print(f"Number of target label: {df["island"].value_counts()}")
print(f"Number of NaN: {df.isna().sum()}")
pd.crosstab(df["species"], df["island"])
```

**Observations:** Chinstrap penguins only appear on Dream island, whereas Gentoo penguins only appear on Biscoe island.

### Visualize variables

```python
sns.countplot(data=df, x="island", hue="species")
plt.show()
```

```python
sns.pairplot(df.dropna(), vars=["bill_length_mm","bill_depth_mm","flipper_length_mm","body_mass_g"], hue="island")
plt.show()
```
